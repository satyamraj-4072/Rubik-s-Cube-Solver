```python
# =====================================================
# RUBIK'S CUBE SOLVER (COLOR INPUT VERSION)
# =====================================================
# Features:
# ✔ User enters cube using COLORS (W, Y, G, B, R, O)
# ✔ Automatically converts to solver format (U, R, F, D, L, B)
# ✔ Validates cube (count + centers)
# ✔ Solves using Kociemba algorithm
# ✔ Shows step-by-step moves
# =====================================================

# Install once (for Google Colab)
# !pip install kociemba matplotlib

import kociemba                      # Solver library
from collections import Counter     # For counting colors


# =====================================================
# COLOR → FACE MAPPING
# =====================================================
# Mapping real cube colors to solver notation

color_to_face = {
    'W': 'U',   # White → Up
    'Y': 'D',   # Yellow → Down
    'G': 'F',   # Green → Front
    'B': 'B',   # Blue → Back
    'R': 'R',   # Red → Right
    'O': 'L'    # Orange → Left
}


# =====================================================
# INPUT FUNCTION (ONE FACE)
# =====================================================
# Takes 3 rows of input and converts colors → letters

def read_one_face_color(face_name):
    print(f"\nEnter {face_name} face using COLORS")
    print("Allowed: W (White), Y (Yellow), G (Green), B (Blue), R (Red), O (Orange)")

    while True:
        face_data = ""

        # Loop for 3 rows
        for i in range(1, 4):
            row = input(f"Row {i}: ").upper().strip().split()

            # Check if exactly 3 inputs
            if len(row) != 3:
                print("❌ Enter exactly 3 colors separated by space")
                break

            # Convert colors to cube notation
            try:
                converted = "".join(color_to_face[c] for c in row)
            except KeyError:
                print("❌ Invalid color! Use only W Y G B R O")
                break

            face_data += converted

        # If valid (9 values), return face
        if len(face_data) == 9:
            return face_data


# =====================================================
# FULL CUBE INPUT
# =====================================================
# Reads all 6 faces

def read_full_cube_color():
    print("\n=== ENTER CUBE USING COLORS ===")

    U = read_one_face_color("UP (White)")
    R = read_one_face_color("RIGHT (Red)")
    F = read_one_face_color("FRONT (Green)")
    D = read_one_face_color("DOWN (Yellow)")
    L = read_one_face_color("LEFT (Orange)")
    B = read_one_face_color("BACK (Blue)")

    # Combine into single cube string
    return U + R + F + D + L + B


# =====================================================
# VALIDATION FUNCTION
# =====================================================
# Checks if cube is valid

def validate_cube(cube):

    # Rule 1: Total stickers must be 54
    if len(cube) != 54:
        return "Cube must have exactly 54 characters"

    # Rule 2: Each face must appear exactly 9 times
    count = Counter(cube)

    for c in "URFDLB":
        if count[c] != 9:
            return f"{c} appears {count[c]} times (should be 9)"

    return "OK"


# =====================================================
# CENTER CHECK (IMPORTANT)
# =====================================================
# Ensures all 6 centers are different

def check_centers(cube):
    centers = [
        cube[4],   # U center
        cube[13],  # R center
        cube[22],  # F center
        cube[31],  # D center
        cube[40],  # L center
        cube[49]   # B center
    ]

    return len(set(centers)) == 6


# =====================================================
# MAIN SOLVER FUNCTION
# =====================================================

def solve_cube():

    # Step 1: Take input
    cube = read_full_cube_color()

    print("\nConverted Cube String:")
    print(cube)

    # Step 2: Validate cube
    status = validate_cube(cube)

    if status != "OK":
        print("\n❌ Invalid Cube:", status)
        return

    # Step 3: Check centers
    if not check_centers(cube):
        print("\n❌ Invalid Cube: Center pieces are incorrect")
        return

    # Step 4: Solve cube
    try:
        solution = kociemba.solve(cube)

        print("\n✅ Solution Found!")
        print("Moves:", solution)

        # Step-by-step output
        print("\nStep-by-step:")
        moves = solution.split()

        for i, move in enumerate(moves):
            print(f"Step {i+1}: {move}")

    except Exception as e:
        print("\n❌ Solver Error:", e)


# =====================================================
# RUN PROGRAM
# =====================================================

solve_cube()
```
